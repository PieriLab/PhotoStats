```python
from sklearn.metrics import adjusted_rand_score

# Suppose we have 5 items
# Ground truth labels
y_true = ['a', 'a', 'b', 'b', 'c']

# Predicted clustering
y_pred =  ['a', 'a', 'b', 'b', 'c']

# Compute Adjusted Rand Index
ari = adjusted_rand_score(y_true, y_pred)
print("Adjusted Rand Index:", ari)
```
