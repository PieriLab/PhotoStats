```python
from sklearn.metrics import adjusted_rand_score, adjusted_mutual_info_score
```

```python
import pandas as pd

# --- Read CSVs ---
predicted = pd.read_csv("/Users/connerbaucom/Desktop/Pieri/CTG/dim_red_comp/PhotoStats/data/clustered/smiles/cartesian/umap/kmeans/benzene/cluster_assignments.csv")
meci_true = pd.read_csv("../../data/meci_classification/benzene/S1S0/meci_labels.csv")

merged_df = pd.merge(predicted, meci_true, on='idx', how='inner')  



print(merged_df.head())
```

```python

predicted_labels = merged_df['cluster'].to_list()
meci_labels = merged_df['meci_type'].to_list()

#print(meci_labels)
#print(predicted_labels)
```

```python


# Compute ARI
ari = adjusted_rand_score(meci_labels, predicted_labels)
print("Adjusted Rand Index:", ari)

# Compute AMI
ami = adjusted_mutual_info_score(meci_labels, predicted_labels)
print("Adjusted Mutual Information:", ami)
```
