```python
from rdkit import Chem
import numpy as np
import pandas as pd 
from pathlib import Path 
import tqdm
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import matplotlib.patches as mpatches
import hdbscan
from matplotlib.colors import ListedColormap
import plotly.graph_objects as go
import sys
from pathlib import Path



repo_root = Path.cwd().parents[2]  
sys.path.insert(0, str(repo_root))

from Photostats.geometry_functions import xyz_to_rdkit_mol, numpy_geom
from Photostats.dim_red import reduce_features
from Photostats.dim_red_comparision import compare_dimensionallity_reduction,create_meci_df
```

```python
raw_spawns_folder = Path('../../../data/aligned_geometries/SeamStress_improper_rotations/benzene/')
mecis = [ 
    Path('../../../data/aligned_geometries/SeamStress/benzene/type2/meci/Type1.xyz'),
    Path('../../../data/aligned_geometries/SeamStress/benzene/type2/meci/Type2.xyz'),
    Path('../../../data/aligned_geometries/SeamStress/benzene/type2/meci/Type3.xyz'),
    
]
```

```python
idx = []
geometries = []

smiles = []

for x in tqdm.tqdm(list(raw_spawns_folder.glob('*'))):
    mol = xyz_to_rdkit_mol(x,total_charge=0)
    smile = Chem.MolToSmiles(mol, canonical=True)
    geom = numpy_geom(mol)
    flattend_geom = geom.flatten()
    geometries.append(flattend_geom)
    idx.append(x.stem)
    smiles.append(smile)


for meci in mecis:
    print(meci)
    mol = xyz_to_rdkit_mol(meci,total_charge=0)
    smile = Chem.MolToSmiles(mol, canonical=True)

    geom = numpy_geom(mol)
    flattend_geom = geom.flatten()
    geometries.append(flattend_geom)
    smiles.append(smile)
    idx.append(meci.stem)

feature_vector = np.array(geometries)





```

```python
csv_file = "../../../data/meci_classification/benzene/S1S0/meci_labels_humanlabels.csv"

# Step 1: Create aligned labeled DataFrame
labeled_df = create_meci_df(idx, csv_file)

# Step 2: Compare dimensionality reduction techniques
results = compare_dimensionallity_reduction(
    feature_vector=feature_vector,
    labeled_df=labeled_df,
    max_dimension=3
)


```

```python
print("UMAP trustworthiness:", results['UMAP']['trustworthiness'])
print("UMAP linear accuracy:", results['UMAP']['linear_accuracy'])
print("TSNE trustworthiness:", results['TSNE']['trustworthiness'])
print("TSNE linear accuracy:", results['TSNE']['linear_accuracy'])
print("PCA trustworthiness:", results['PCA']['trustworthiness'])
print("PCA linear accuracy:", results['PCA']['linear_accuracy'])
```
