```python
import csv
from pathlib import Path

# === CONFIG ===
csv_path = Path("/Users/connerbaucom/Desktop/Pieri/CTG/dim_red_comp/PhotoStats/data/meci_classification/benzene/S1S0/meci_labels_0.05_threshold_humanlabels.csv")          # Path to your CSV
directory_path = Path("/Users/connerbaucom/Desktop/Pieri/CTG/dim_red_comp/PhotoStats/data/aligned_geometries/seamstress_reflection/benzene/1_2/spawns")   # Directory containing files to check
csv_column_name = "idx"         # Column in CSV that contains valid stems
# =================

def get_valid_stems(csv_file, column_name):
    valid = set()
    with csv_file.open(newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            stem = row[column_name].strip()
            if stem:
                valid.add(stem)
    return valid

def delete_unmatched_files(directory, valid_stems):
    for file_path in directory.iterdir():
        if file_path.is_file():
            if file_path.stem not in valid_stems:
                print(f"Deleting: {file_path.name}")
                file_path.unlink()

if __name__ == "__main__":
    valid_stems = get_valid_stems(csv_path, csv_column_name)
    delete_unmatched_files(directory_path, valid_stems)
    print("Done.")
```
