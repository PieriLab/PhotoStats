```python
import os
import numpy as np
import pandas as pd
from pathlib import Path
from ase.io import read, write
from io import StringIO
import rmsd as  rmsdlib
import tqdm
```

```python
def align_to_ref(ref_file, target_file):
    """
    Returns (RMSD, ASE.Atoms aligned structure)
    """
    cmd = f"-e --reorder-method inertia-hungarian --use-reflections -p {ref_file} {target_file}"
    out = rmsdlib.main(cmd.split())

    lines = out.strip().splitlines()

    try:
        rmsd_val = float(lines[0].split()[1])
        xyz_block = "\n".join(lines[1:])
    except Exception:
        rmsd_val = np.nan
        xyz_block = "\n".join(lines)
        atoms = read(StringIO(xyz_block), format="xyz")

    return rmsd_val, atoms
```

```python
unaligned_folder = Path('/Users/connerbaucom/Desktop/Pieri/CTG/dim_red_comp/PhotoStats/data/raw_geometries/spawn/ethylene/')
reference = Path('/Users/connerbaucom/Desktop/Pieri/CTG/dim_red_comp/PhotoStats/data/raw_geometries/meci/ethylene/0000_2.xyz')

ase_geom = []
names = []
for x in tqdm.tqdm(list(unaligned_folder.glob('*'))):
    rmsd, atoms = align_to_ref(reference,x)

    ase_geom.append(atoms)
    name = x.stem
    names.append(name)
```
