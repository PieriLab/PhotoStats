```python
import pandas as pd

# File names
file1 = "/Users/connerbaucom/Desktop/Pieri/CTG/dim_red_comp/PhotoStats/data/AIMS_spawn_energies.csv"      # first CSV (IC_Spawn,...)
file2 = "/Users/connerbaucom/Desktop/Pieri/CTG/dim_red_comp/PhotoStats/data/meci_classification/benzene/S2S1/meci_labels_0.05_threshold_humanlabels.csv"   # second CSV (idx,meci_type)
output = "/Users/connerbaucom/Desktop/Pieri/CTG/dim_red_comp/PhotoStats/data/merged_csv/benzene_s2s1.csv"


df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

merged = df1.merge(
    df2,
    left_on="IC_Spawn",
    right_on="idx",
    how="inner",
    sort=False
).drop(columns=["idx"])

merged.to_csv(output, index=False)
```
